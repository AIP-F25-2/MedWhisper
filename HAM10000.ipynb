```python
import os, random, shutil, pathlib, json
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm

import torch, torch.nn as nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from torchvision import datasets, transforms
import timm

from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Grad-CAM availability flags
HAS_PTCAM = False
HAS_TORCHCAM = False
try:
    from pytorch_grad_cam import GradCAM
    from pytorch_grad_cam.utils.image import show_cam_on_image
    HAS_PTCAM = True
except Exception:
    try:
        from torchcam.methods import GradCAM as TorchGradCAM
        HAS_TORCHCAM = True
    except Exception:
        pass

# Repro & training settings
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

IMG_SIZE    = 224
BATCH_SIZE  = 32
EPOCHS      = 12        # increase later if you want
LR          = 3e-4
CONF_THRESH = 0.6
DEVICE      = "cuda" if torch.cuda.is_available() else "cpu"

# HAM10000 7 classes
CLASSES = ["akiec","bcc","bkl","df","mel","nv","vasc"]

print("Device       :", DEVICE)
print("CAM backend  :", "pytorch-grad-cam" if HAS_PTCAM else ("torchcam" if HAS_TORCHCAM else "none"))
```

```python
# Folder that contains: HAM10000_metadata.csv, HAM10000_images_part_1, HAM10000_images_part_2
ROOT_DIR = r"C:\Users\Nitro\OneDrive\Desktop\skin-dermotology\dataset"  # <-- EDIT if your path differs

META_CSV = os.path.join(ROOT_DIR, "HAM10000_metadata.csv")
PART1    = os.path.join(ROOT_DIR, "HAM10000_images_part_1")
PART2    = os.path.join(ROOT_DIR, "HAM10000_images_part_2")

for p in [META_CSV, PART1, PART2]:
    assert os.path.exists(p), f"Missing: {p}"

IMG_DIRS = [PART1, PART2]
print("ROOT_DIR:", ROOT_DIR)
```

```python
OUT_SPLIT = "data/split"

# Create folders
for split in ["train","val","test"]:
    for c in CLASSES:
        pathlib.Path(os.path.join(OUT_SPLIT, split, c)).mkdir(parents=True, exist_ok=True)

# Load metadata and keep only our 7 classes
meta = pd.read_csv(META_CSV)
meta = meta[meta["dx"].isin(CLASSES)].copy()

# Map image_id → absolute path
def path_for(img_id: str):
    name = img_id + ".jpg"
    for d in IMG_DIRS:
        p = os.path.join(d, name)
        if os.path.exists(p): return p
    return None

meta["img_path"] = meta["image_id"].apply(path_for)
meta = meta.dropna(subset=["img_path"]).reset_index(drop=True)

# Split by lesion_id (prevents patient/image leakage)
lesions = meta["lesion_id"].dropna().unique().tolist()
random.shuffle(lesions)
n = len(lesions)
train_ids = set(lesions[:int(0.70*n)])
val_ids   = set(lesions[int(0.70*n):int(0.85*n)])
test_ids  = set(lesions[int(0.85*n):])

def which_split(lid):
    return "train" if lid in train_ids else ("val" if lid in val_ids else "test")

meta["split"] = meta["lesion_id"].apply(which_split)

# Copy files into data/split/<split>/<class>/image.jpg
for _, r in tqdm(meta.iterrows(), total=len(meta), desc="Copying → data/split"):
    dst = os.path.join(OUT_SPLIT, r["split"], r["dx"], os.path.basename(r["img_path"]))
    if not os.path.exists(dst):
        shutil.copy2(r["img_path"], dst)

print(meta["split"].value_counts())
print("Per-class counts:\n", meta["dx"].value_counts())
```

```python
from collections import Counter

# Transforms
train_tfms = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.8,1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(0.1,0.1,0.1,0.05),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
])
eval_tfms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
])

# Datasets
train_ds = datasets.ImageFolder(os.path.join(OUT_SPLIT,"train"), transform=train_tfms)
val_ds   = datasets.ImageFolder(os.path.join(OUT_SPLIT,"val"),   transform=eval_tfms)
test_ds  = datasets.ImageFolder(os.path.join(OUT_SPLIT,"test"),  transform=eval_tfms)
classes  = train_ds.classes
print("Classes:", classes)

# Weighted sampler to reduce class imbalance
cnt = Counter([y for _, y in train_ds.samples])
class_count = np.array([cnt[i] for i in range(len(classes))])
weights = 1.0 / np.clip(class_count, 1, None)
sample_w = np.array([weights[y] for _, y in train_ds.samples], dtype=np.float32)
sampler  = WeightedRandomSampler(sample_w, num_samples=len(sample_w), replacement=True)

# Loaders
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler, num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

# Sanity check
xb, yb = next(iter(train_loader))
print("Batch:", xb.shape, yb.shape)
```

```python
from tqdm import tqdm

model = timm.create_model("efficientnet_b0", pretrained=True, num_classes=len(classes)).to(DEVICE)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=LR)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

def run_epoch(loader, train=False):
    model.train(train)
    total=correct=0; losses=[]
    for x,y in tqdm(loader, leave=False):
        x,y = x.to(DEVICE), y.to(DEVICE)
        with torch.set_grad_enabled(train):
            logits = model(x); loss = criterion(logits, y)
        if train:
            optimizer.zero_grad(); loss.backward(); optimizer.step()
        total += y.size(0)
        correct += (logits.argmax(1)==y).sum().item()
        losses.append(loss.item())
    return float(np.mean(losses)), correct/total

best_val = 0.0
for epoch in range(1, EPOCHS+1):
    tr_loss, tr_acc = run_epoch(train_loader, train=True)
    va_loss, va_acc = run_epoch(val_loader,   train=False)
    scheduler.step()
    print(f"Epoch {epoch:02d} | Train {tr_acc:.3f}/{tr_loss:.3f}  Val {va_acc:.3f}/{va_loss:.3f}")
    if va_acc > best_val:
        best_val = va_acc
        torch.save({"model":model.state_dict(),"classes":classes}, "best_model.pt")
        print("  ✓ saved best_model.pt")
```

```python
ckpt = torch.load("best_model.pt", map_location=DEVICE)
model.load_state_dict(ckpt["model"]); model.eval()
classes = ckpt["classes"]

def collect(loader):
    outs=[]; ys=[]
    with torch.no_grad():
        for x,y in loader:
            outs.append(model(x.to(DEVICE)).cpu()); ys.append(y)
    return torch.cat(outs), torch.cat(ys)

logits, y = collect(test_loader)
probs = logits.softmax(1).numpy()
y_true = y.numpy()
y_pred = probs.argmax(1)

print("\nClassification report:\n",
      classification_report(y_true, y_pred, target_names=classes, digits=3))

try:
    y_oh = np.eye(len(classes))[y_true]
    auroc = roc_auc_score(y_oh, probs, average="macro", multi_class="ovr")
    print("Macro AUROC:", round(auroc,3))
except Exception as e:
    print("AUROC note:", e)

print("Confusion matrix:\n", confusion_matrix(y_true, y_pred))
```

```python
def predict_image(image_path, conf_thresh=0.6):
    ckpt = torch.load("best_model.pt", map_location=DEVICE)
    classes = ckpt["classes"]
    mdl = timm.create_model("efficientnet_b0", pretrained=False, num_classes=len(classes)).to(DEVICE)
    mdl.load_state_dict(ckpt["model"]); mdl.eval()

    tfm = transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
    ])
    img = Image.open(image_path).convert("RGB")
    x = tfm(img).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        prob = mdl(x).softmax(1).squeeze(0).cpu().numpy()
    idx = int(prob.argmax()); conf = float(prob[idx])
    return {
        "prediction": "ABSTAIN" if conf < conf_thresh else classes[idx],
        "confidence": conf,
        "probs": {c: float(p) for c,p in zip(classes, prob)}
    }

# Example usage:
# import pathlib
# sample = next(iter(pathlib.Path("data/split/test").rglob("*.jpg")))
# print(sample, predict_image(str(sample)))
```

```python
if HAS_PTCAM:
    os.makedirs("outputs/cam_samples", exist_ok=True)

    # inverse normalization for display
    inv = transforms.Normalize(
        mean=[-m/s for m,s in zip([0.485,0.456,0.406],[0.229,0.224,0.225])],
        std=[1/s for s in [0.229,0.224,0.225]]
    )

    target_layer = model.conv_head if hasattr(model, "conv_head") else list(model.children())[-1]
    cam = GradCAM(model=model, target_layers=[target_layer], use_cuda=False)

    saved=0
    for x,_ in test_loader:
        x = x.to(DEVICE)
        with torch.no_grad():
            logits = model(x)
            p = logits.softmax(1)
            conf, pred = p.max(1)
        for i in range(min(8, x.size(0))):
            img = x[i].detach().cpu()
            np_img = inv(img).clamp(0,1).permute(1,2,0).numpy().astype(np.float32)
            label = "ABSTAIN" if conf[i].item()<CONF_THRESH else classes[pred[i].item()]
            grayscale_cam = cam(input_tensor=x[i].unsqueeze(0), targets=None)[0]
            overlay = show_cam_on_image(np_img, grayscale_cam, use_rgb=True)
            Image.fromarray((overlay*255).astype(np.uint8)).save(f"outputs/cam_samples/cam_{saved}_{label}.jpg")
            saved += 1
        if saved>=16: break

    print("Saved CAMs → outputs/cam_samples/ (pytorch-grad-cam)")
else:
    print("pytorch-grad-cam not available; use the torchcam cell if installed.")
```

```python
if HAS_TORCHCAM:
    import matplotlib.pyplot as plt
    os.makedirs("outputs/cam_samples", exist_ok=True)

    # choose last Conv2d as target
    target_layer = None
    for m in reversed(list(model.modules())):
        if isinstance(m, torch.nn.Conv2d):
            target_layer = m; break
    cam_extractor = TorchGradCAM(model, target_layer=target_layer)

    inv = transforms.Normalize(
        mean=[-m/s for m,s in zip([0.485,0.456,0.406],[0.229,0.224,0.225])],
        std=[1/s for s in [0.229,0.224,0.225]]
    )

    saved=0
    model.eval()
    for x,_ in test_loader:
        x = x.to(DEVICE)
        scores = model(x)
        for i in range(min(8, x.size(0))):
            cls = int(scores[i].argmax().item())
            cam_map = cam_extractor(class_idx=cls, scores=scores)[i].detach().cpu()  # HxW
            img = x[i].detach().cpu()
            img_disp = inv(img).clamp(0,1).permute(1,2,0).numpy()
            cam_np = torch.nn.functional.interpolate(
                cam_map.unsqueeze(0).unsqueeze(0),
                size=(img_disp.shape[0], img_disp.shape[1]),
                mode="bilinear", align_corners=False
            ).squeeze().numpy()
            plt.figure(figsize=(3,3)); plt.axis('off')
            plt.imshow(img_disp); plt.imshow(cam_np, alpha=0.4); plt.title(CLASSES[cls])
            plt.savefig(f"outputs/cam_samples/cam_{saved}_{CLASSES[cls]}.jpg",
                        bbox_inches='tight', pad_inches=0)
            plt.close(); saved += 1
        if saved>=16: break

    print("Saved CAMs → outputs/cam_samples/ (torchcam)")
else:
    print("torchcam not available; CAMs skipped.")
```

```python
# --- HAM10000 leakage-safe split (notebook version) ---
import os, sys, random, shutil, pathlib
from typing import Optional, List
import pandas as pd
from tqdm import tqdm

CLASSES = ["akiec","bcc","bkl","df","mel","nv","vasc"]
REQUIRED_COLS = {"image_id","lesion_id","dx"}

def find_dir_case_insensitive(root: str, name: str) -> Optional[str]:
    for entry in os.listdir(root):
        p = os.path.join(root, entry)
        if os.path.isdir(p) and entry.lower() == name.lower():
            return p
    return None

def path_for(img_id: str, img_dirs: List[str]) -> Optional[str]:
    name = img_id + ".jpg"
    for d in img_dirs:
        p = os.path.join(d, name)
        if os.path.exists(p):
            return p
    return None

def safe_copy(src: str, dst: str):
    if os.path.exists(dst):  # no double copy
        return
    pathlib.Path(os.path.dirname(dst)).mkdir(parents=True, exist_ok=True)
    shutil.copy2(src, dst)

# --- EDIT THIS to your dataset folder ---
ROOT_DIR = r"C:\Users\Nitro\OneDrive\Desktop\skin-dermotology\dataset"
OUT_SPLIT = "data/split"
SEED = 42
# ---------------------------------------

# Validate inputs
meta_csv = os.path.join(ROOT_DIR, "HAM10000_metadata.csv")
if not os.path.exists(meta_csv):
    raise FileNotFoundError(f"CSV not found: {meta_csv}")
p1 = find_dir_case_insensitive(ROOT_DIR, "HAM10000_images_part_1")
p2 = find_dir_case_insensitive(ROOT_DIR, "HAM10000_images_part_2")
if not p1 or not p2:
    raise FileNotFoundError("Image folders HAM10000_images_part_1/2 not found (case-insensitive).")
IMG_DIRS = [p1, p2]
print("[OK] CSV:", meta_csv)
print("[OK] Images:", p1)
print("[OK] Images:", p2)

# Read + filter
meta = pd.read_csv(meta_csv)
missing = REQUIRED_COLS - set(meta.columns)
if missing:
    raise ValueError(f"CSV missing columns: {sorted(missing)}")

meta = meta[meta["dx"].isin(CLASSES)].copy()
meta["img_path"] = meta["image_id"].astype(str).apply(lambda x: path_for(x, IMG_DIRS))
before = len(meta)
meta = meta.dropna(subset=["img_path"]).reset_index(drop=True)
if len(meta) < before:
    print(f"[WARN] skipped {before - len(meta)} rows with missing image files.")

# Split by lesion_id
meta = meta.dropna(subset=["lesion_id"]).reset_index(drop=True)
lesions = meta["lesion_id"].astype(str).unique().tolist()
random.seed(SEED); random.shuffle(lesions)
n = len(lesions)
cut1, cut2 = int(0.70*n), int(0.85*n)
train_ids, val_ids, test_ids = set(lesions[:cut1]), set(lesions[cut1:cut2]), set(lesions[cut2:])
meta["split"] = meta["lesion_id"].astype(str).apply(
    lambda lid: "train" if lid in train_ids else ("val" if lid in val_ids else "test")
)

# Create dirs then copy
for split in ["train","val","test"]:
    for c in CLASSES:
        pathlib.Path(os.path.join(OUT_SPLIT, split, c)).mkdir(parents=True, exist_ok=True)

print(f"[INFO] copying images → {OUT_SPLIT} ...")
for _, r in tqdm(meta.iterrows(), total=len(meta)):
    dst = os.path.join(OUT_SPLIT, r["split"], r["dx"], os.path.basename(r["img_path"]))
    safe_copy(r["img_path"], dst)

# Summaries
print("\n[SUMMARY] split sizes:")
print(meta["split"].value_counts().sort_index())
print("\n[SUMMARY] per-class counts:")
print(meta["dx"].value_counts().sort_index())

# Leakage check: lesion_id must not appear in multiple splits
cross = meta.groupby("lesion_id")["split"].nunique().reset_index(name="n_splits")
offenders = cross[cross["n_splits"] > 1]
if not offenders.empty:
    raise RuntimeError("Leakage detected: some lesion_id appear in multiple splits.")
print("\n[OK] leakage check passed ✅")
```
